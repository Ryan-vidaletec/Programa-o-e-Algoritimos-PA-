```python
#Exercício 1 — Contagem

a = int(input("Digite um número: "))
b = 1

while b <= a:
  print(b)
  b = b + 1


```

```python
#Exercício 2 — Contagem regressiva

a = int(input("Digite um número: "))
contador = a

while contador >= 0:
  print(contador)
  contador = contador - 1
```

```python
#Exercício 3 — Números pares

contador = 0

while contador <= 20:
  print(contador)
  contador = contador + 2
```

```python
#Exercício 4 — Soma dos números

numero = int(input("Digite um número: "))

soma = 0
contador = 1

while contador <= numero:
    soma = soma + contador
    contador = contador + 1

print("Soma:", soma)
```

```python
#Exercício 5 — Tabuada

a = int(input("Digite um número: "))
contador = 1
mult = 1

while contador <= 10:
  print(f"{a} X {mult} = ({a * contador})")
  contador = contador + 1
  mult = mult + 1
```

```python
#Exercício 6 — Senha

senha = 1234

a = int(input("Digite a senha: "))

while a != senha:
  print("Senha incorreta! Tente Novamente: ")
  a = int(input("Digite a senha: "))

if a == senha:
  print("Acesso liberado! ")
```

```python
#Exercício 7 — Somando até digitar zero

soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
    soma = soma + numero
    numero = int(input("Digite um número: "))

print("Soma:", soma)
```

```python
#Exercício 8 — Maior número

maior = 0

numero = int(input("Digite um número: "))

while numero != 0:
    if numero > maior:
        maior = numero

    numero = int(input("Digite um número: "))

print("Maior número:", maior)
```

```python
#Exercício 9 — Média das notas

soma = 0
contador = 0

while contador < 5:
    nota = float(input("Digite a nota: "))
    soma = soma + nota
    contador = contador + 1

media = soma / 5

print("Média:", media)
```

```python
saldo = 1000

while True:
    print("\n1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        valor = float(input("Digite o valor do depósito: "))
        saldo = saldo + valor
        print("Depósito realizado!")

    elif opcao == 2:
        valor = float(input("Digite o valor do saque: "))

        if valor <= saldo:
            saldo = saldo - valor
            print("Saque realizado!")
        else:
            print("Saldo insuficiente!")

    elif opcao == 3:
        print("Saldo: R$", saldo)

    elif opcao == 4:
        print("Obrigado por usar o caixa eletrônico!")
        break

    else:
        print("Opção inválida!")
```
