```python
#Exercício 1 – Criando variáveis

nome = "Vidal"
idade = 16
altura = 1.79
nota = 6.7
cidade = "Rua Belterra, 291, Santo Amaro, SP"

print(nome)
print(type(nome))
print("---------------X---------------")
print(idade)
print(type(idade))
print("---------------X---------------")
print(altura)
print(type(altura))
print("---------------X---------------")
print(nota)
print(type(nota))
print("---------------X---------------")
print(cidade)
print(type(cidade))
print("---------------X---------------")

```

```python
#Exercício 2 – Contas básicas

n1 = 7.5
n2 = 8.0
n3 = 6.5

média = (n1 + n2 + n3) /3

print(f"A sua nota 1 foi = {n1}, sua nota 2 foi = {n2}, e sua nota 3 foi = {n3}, fechando com a média de: {média}")
```

```python
#Exercício 3 – Conversões

texto1 = "20"
texto2 = "3.5"
numero = 8


#A)
a = int(texto1) + 10

#B)
b = float(texto2) * 2

#C)
c = str(numero) + " anos"


print(a)
print(b)
print(c)
```

```python
#Exercício 4 – Expressões aritméticas

# O dobro de 15
print(f"O dobro de 15 é igual a: {15*2}")

print("----------X----------")

#• A metade de 90
print(f"A metade de 90 é igual a: {90/2}")

print("----------X----------")

#• O resto da divisão de 17 por 5
print(f"O resto da divisão de 17 por 5 é igual a: {17%5}")

print("----------X----------")

#• 2 elevados à potência 4
print(f"O 2 elevado à potência 4 é igual a: {2**4}")

print("----------X----------")

#• A média de 10, 8 e 6
print(f"A média de 10, 8 e 6 é igual a: {(10 + 8 + 6) / 3}")
```

```python
#Exercício 5 – Comparações (relacionais)

idade = 16
nota = 8.5
preço = 45.0


print(f"idade >= 18 {idade >= 18}")
print(f"nota > 7 {nota > 7}")
print(f"preço <= 50 {preço <= 50}")
print(f"idade == 16 {idade == 16}")
print(f"nota != 10 {nota != 10}")
```

```python
#Exercício 6 – Situação prática

valor = 80
desconto = valor * 0.10
vf = valor - desconto

print(f"O valor do desconto é igual a: {desconto}")
print("----------X----------")
print(f"O valor final a pagar é igual a: {vf}")
print("----------X----------")
print(f"O valor final a pagar é menor que 75 {vf < 75}")
```
