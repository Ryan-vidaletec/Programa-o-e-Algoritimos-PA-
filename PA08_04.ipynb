```python
# Fazer um código que identifica se o número é par ou impar

a = int(input("Digite um número:"))


if a % 2 == 0:
  print(f"O número {a} é par!")

else:
  print(f"O número {a} é impar!")
```

```python
# Escrever um programa que recebe 04 notas do aluno, calcula a média e mostra se o aluno foi aprovado, reprovado, ou recuperação


# Entrada

print("---------------------------------------------------------------------------------")
nome = str(input("Digite seu nome: "))
print("---------------------------------------------------------------------------------")
print(f"Digite suas notas abaixo, notas de 0 a 10:")
print("---------------------------------------------------------------------------------")

nota1 = float (input("Digite a nota do 1 bimestre: "))
if nota1 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

nota2 = float (input("Digite a nota do 2 bimestre: "))
if nota2 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

nota3 = float (input("Digite a nota do 3 bimestre: "))
if nota3 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

nota4 = float (input("Digite a nota do 4 bimestre: "))
if nota4 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")


# Processamento

resultado = (nota1 + nota2 + nota3 + nota4) / 4

if nota1 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

elif resultado >= 7:
  print(f"A média das notas do aluno {nome} foi: {resultado}, você foi aprovado!")

elif resultado >= 5:
  print(f"A média das notas do aluno {nome} foi: {resultado}, você está de recuperação!")

else:
  print(f"A média das notas do aluno {nome} foi: {resultado}, você foi reprovado!")
```

```python
# Escrever um programa que recebe os valores de 3 lados e verificar se é um triângulo válido
# Classificar o tipo de de triângulo
# Verificar se o triângulo é válido
# A soma das medidas de quaisquer dois lados deve ser sempre maior que a medida do terceiro lado.
# Classificar tipos de triângulos
# Equilatero - todos os lados iguais
# Isoceles - possui 2 lados iguais
# Escaleno - Todos os lados diferentes

l1 = float(input("Digite o 1 lado:"))
l2 = float(input("Digite o 2 lado:"))
l3 = float(input("Digite o 3 lado:"))

if l1 + l2 > l3 and l3 + l2 > l1 and l1 + l3 > l2:
  print("Esse é um triângulo válido!")

else:
  print("Esse é um triângulo inválido!")

if l1 == l2 and l2 == l3:
  print("É um triângulo Equilátero!")

elif l2 == l1 or l3 == l2 or l3 == l1:
  print("É um triângulo Isósceles!")

else:
  print("É um triãngulo Escaleno!")
```
