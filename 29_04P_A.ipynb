```python
# Escrever um programa que o usuário digita uma palavra e o programa verifica se a palavra é um palíndromo e imprime

a = str(input("Digite uma palavra: "))
invertida = ""

for i in (a):
  invertida = i + invertida

if invertida == a:
  print(f"A palavra {a} invertida é {invertida} e é um Palíndromo! ")

else:
  print(f"A palavra {a} invertida é {invertida} e não é um Palíndromo! ")
```
