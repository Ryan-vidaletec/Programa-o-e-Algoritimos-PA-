```python
# Par ou ímpar, peça um número ao usuário e informe se ele é par ou ímpar.

print("NÚMERO PAR E ÍMPAR")

n1 = int(input("Digite um número:"))

if n1 % 2 == 0:
  print(f"O número {n1} é par!")

else:
  print(f"O número {n1} é ímpar!")
```

```python
# Maior de dois números, solicite dois números e mostre qual é o maior (ou se são iguais).

print("NÚMERO MAIOR, MENOR OU IGUAL")

a = float(input("Digite o primeiro número:"))
b = float(input("Digite o segundo número:"))

if a > b:
  print(f"O número {a} é maior que {b}!")

elif b > a:
  print(f"O número {b} é maior que {a}")

else:
  print("Os números são iguais!")
```

```python
# Aprovação de aluno, peça a nota de um aluno e informe:
# Aprovado (nota ≥ 7) Recuperação (5 ≤ nota < 7) Reprovado (nota < 5)

print(f"Digite suas notas abaixo, notas de 0 a 10:")
print("---------------------------------------------------------------------------------")

nota1 = float (input("Digite a nota do 1 bimestre: "))
if nota1 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

nota2 = float (input("Digite a nota do 2 bimestre: "))
if nota2 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

nota3 = float (input("Digite a nota do 3 bimestre: "))
if nota3 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

nota4 = float (input("Digite a nota do 4 bimestre: "))
if nota4 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")


# Processamento

resultado = (nota1 + nota2 + nota3 + nota4) / 4

if nota1 > 10:
  print("Digite uma nota válida (notas de 0 a 10)")

elif resultado >= 7:
  print(f"A média das notas foi: {resultado}, você foi aprovado!")

elif resultado >= 5:
  print(f"A média das notas foi: {resultado}, você está de recuperação!")

else:
  print(f"A média das notas foi: {resultado}, você foi reprovado!")

```

```python
# Classificação de idade, receba a idade e classifique: Criança (0–12) Adolescente (13–17) Adulto (18–59) Idoso (60+)

id = int(input("Digite sua idade:"))

if id < 13:
  print("Você foi classificado(a) como: Criança!")

if id < 18 and  id > 12:
  print("Você foi classificado(a) como: Adolescente!")

if id < 60 and id > 17:
  print("Você foi classificado(a) como: Adulto!")

if id > 59:
  print("Você foi classificado(a) como : Idoso!")
```

```python
# Positivo, negativo ou zero, peça um número e diga se ele é positivo, negativo ou zero.

print("NÚMERO POSITIVO, NEGATIVO OU ZERO")

a = int(input("Digite um número:"))

if a == 0:
  print(f"O número {a} é zero!")

if a > 0:
  print(f"O número {a} é positivo!")

if a < 0:
  print(f"O número {a} é negativo!")
```

```python
# Calculadora simples, Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o resultado.

print(" Opção - 1 = SOMA")
print(" Opção - 2 = SUBTRAÇÃO")
print(" Opção - 3 = MULTIPLICAÇÃO")
print(" Opção - 4 = DIVISÃO")
print("")

opcao = int (input("Digite a opção desejada:"))

# Entrada

a = float (input("Digite o primeiro número:"))
b = float (input("Digite o segundo número:"))

if opcao == 1:
  print(f"A soma de {a} mais {b} é igual a {a + b}.")

elif opcao == 2:
  print(f"A subtração de {a} menos {b} é igual a {a - b}.")

elif opcao == 3:
  print(f"A multiplicação de {a} vezes {b} é igual a {a * b}.")

elif opcao == 4:
  print(f"A divisão de {a} mais {b} é igual a {a / b}")

else:
  print("Digite uma opção válida!")
```

```python
# Dia da semana, peça um número de 1 a 7 e mostre o dia correspondente (ex: 1 = domingo).

dia = int(input("Digite o número do dia da semana de 1 a 7:"))

if dia == 1:
  print("Domingo")

if dia == 2:
  print("Segunda")

if dia == 3:
  print("Terça")

if dia == 4:
  print("Quarta")

if dia == 5:
  print("Quinta")

if dia == 6:
  print("Sexta")

if dia == 7:
  print("Sábado")

if dia > 7 or dia < 1:
  print("Digite um dia válido!")
```

```python
# Ano bissexto, solicite um ano e informe se ele é bissexto ou não.

print("ANO BISSEXTO")
print("")
ano = int(input("Digite o ano:"))

if ano % 4 == 0 or ano % 400 == 0:
  print("O ano é bissexto!")

else:
  print("O ano não é bissexto!")
```

```python
# Desconto em compras, peça o valor de uma compra: ≥ 100: 10% de desconto ≥ 50 e < 100: 5% de desconto < 50: sem desconto

print("DESCONTOS")

valor = int(input("Digite o valor da compra:"))

if valor > 100:
  print("Desconto de 10%!")

if valor > 50 and valor < 101:
  print("Desconto de 5%!")

if valor < 51:
  print("Sem desconto!")
```
