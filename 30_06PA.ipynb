```python
# Cofre digital

# Um cofre possui um código composto  por 5 números inteiros.

# Desenvolva um programa que:

# Leia os 5 números utilizando um laço de repetição.

# Ao final informar:

# 1- Quantos números são pares.
# 2- Quantos números são ímpares.
# 3- O maior número.
# 4- A soma de todos os números.
# 5- A média dos números.
# 6- O menor número.
# 7- Quantos números são maiores que a média.


pares = 0
impares = 0
soma = 0
maiores_media = 0

n1 = int(input("Digite o 1º número: "))
n2 = int(input("Digite o 2º número: "))
n3 = int(input("Digite o 3º número: "))
n4 = int(input("Digite o 4º número: "))
n5 = int(input("Digite o 5º número: "))

soma = n1 + n2 + n3 + n4 + n5
media = soma / 5

if n1 % 2 == 0:
    pares += 1
else:
    impares += 1

if n2 % 2 == 0:
    pares += 1
else:
    impares += 1

if n3 % 2 == 0:
    pares += 1
else:
    impares += 1

if n4 % 2 == 0:
    pares += 1
else:
    impares += 1

if n5 % 2 == 0:
    pares += 1
else:
    impares += 1

maior = n1
if n2 > maior:
    maior = n2
if n3 > maior:
    maior = n3
if n4 > maior:
    maior = n4
if n5 > maior:
    maior = n5

menor = n1
if n2 < menor:
    menor = n2
if n3 < menor:
    menor = n3
if n4 < menor:
    menor = n4
if n5 < menor:
    menor = n5

if n1 > media:
    maiores_media += 1
if n2 > media:
    maiores_media += 1
if n3 > media:
    maiores_media += 1
if n4 > media:
    maiores_media += 1
if n5 > media:
    maiores_media += 1

print("Pares:", pares)
print("Ímpares:", impares)
print("Maior número:", maior)
print("Soma:", soma)
print("Média:", media)
print("Menor número:", menor)
print("Números maiores que a média:", maiores_media)
```

```python
# Número espelho

# Solicite um número inteiro positivo

# Inverta o número
# Informe se ele é um palíndromo


num = int(input("Digite um número inteiro positivo: "))

original = num
invertido = 0

while num > 0:
    resto = num % 10
    invertido = invertido * 10 + resto
    num = num // 10

if original == invertido:
    print("O número é um palíndromo")
else:
    print("O número não é um palíndromo")
```
