```python
# 1- Contagem simples, o usuário digita um número e o programa imprime do zero até o número

num = int(input("Digite um número: "))

for i in range(num):
  print(i + 1)
```

```python
# 2- Contagem regressiva, o usuário digita um número e o progresso e o programa imprime do número até zero

num = int(input("Digite um número: "))

for i in range(num + 1):
  print(num - i)
```

```python
# 3- Números pares, o usuário digita um número e o programa imprime os números pares

num = int(input("Digite um número: "))

for i in range(num + 1):
  if i % 2 == 0:
    print(i)
```

```python
# 4- Números impares, o usuário digita um número e o programa imprime os números impares

num = int(input("Digite um número: "))

for i in range(num + 1):
  if i % 2 == 1:
    print(i)
```

```python
# 5- Soma, o usuário digita um número e o programa soma os números do zero até o número

num = int(input("Digite um número: "))

soma = 0
for i in range(num + 1):
  soma += i

print(f"A soma dos números de 0 até {num} é: {soma}")
```

```python
# 6- Tabuada, o usuário digita um número e o programa imprime a tabuada do número

num = int(input("Digite um número: "))

for i in range(11):
  print(f"{num} X {i} = {num * i}")
```

```python
# 11- Pirâmide Numérica Simétrica
# Faça um programa que receba um número N e imprima uma pirâmide perfeitamente centralizada
# Exemplo para N = 5:

#          1
#         212
#        32123
#       4321234
#      543212345

```

```python
# 11- Pirâmide Numérica Simétrica (versão com restrições de palavras-chave)
# Programa que recebe N e imprime uma pirâmide centralizada, usando só o básico.

n = int(input("Digite um número para a pirâmide: "))

for linha in range(1, n + 1):

    parte_esquerda = ""
    for num_desc in range(linha, 0, -1):
        parte_esquerda += str(num_desc)


    parte_direita = ""

    if linha > 1:
        for num_asc in range(2, linha + 1):
            parte_direita += str(num_asc)


    sequencia_numerica = parte_esquerda + parte_direita


    largura_total = n * 2 - 1
    espacos_laterais = (largura_total - len(sequencia_numerica)) // 2


    print(" " * espacos_laterais + sequencia_numerica)
```
