```python
# Escrever um programa que recebe 04 notas do aluno, calcula a média e mostra se o aluno foi aprovado, reprovado, ou recuperação


# Entrada


print("---------------------------------------------------------------------------------")
print(f"Digite suas notas abaixo, notas de 0 a 10:")
print("---------------------------------------------------------------------------------")

nota1 = float (input("Digite a nota do 1 bimestre: "))
nota2 = float (input("Digite a nota do 2 bimestre: "))
nota3 = float (input("Digite a nota do 3 bimestre: "))
nota4 = float (input("Digite a nota do 4 bimestre: "))

# Processamento

resultado = (nota1 + nota2 + nota3 + nota4) / 4

if resultado >= 7:

  print(f"A média das suas notas foi {resultado}, você foi aprovado!")

elif resultado >= 5:

  print(f"A média das suas notas foi {resultado}, você está de recuperação!")

else:

  print(f"A média das suas notas foi {resultado}, você foi reprovado!")
```

```python
# Fazer uma calculadora simples com as 04 operações

print("---------------MENU---------------")
print("")
print(" Opção - 1 = SOMA")
print(" Opção - 2 = SUBTRAÇÃO")
print(" Opção - 3 = MULTIPLICAÇÃO")
print(" Opção - 4 = DIVISÃO")
print("")

opcao = int (input("Digite a opção desejada:"))

# Entrada

a = float (input("Digite o primeiro número:"))
b = float (input("Digite o segundo número:"))

if opcao == 1:
  print(f"A soma de {a} mais {b} é ingual a {a + b}.")

elif opcao == 2:
  print(f"A subtração de {a} menos {b} é ingual a {a - b}.")

elif opcao == 3:
  print(f"A multiplicação de {a} vezes {b} é ingual a {a * b}.")

elif opcao == 4:
  print(f"A divisão de {a} mais {b} é ingual a {a / b}")

else:
  print("Digite uma opção válida!")
```
