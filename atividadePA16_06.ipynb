```python
#QUESTÃO 02 – Soma dos Números (2 pontos)
#-----------------------------------------------------------

#Solicite ao usuário um número inteiro positivo.

#Utilizando um laço de reprodução, calcule e exiba a soma de
#todos os números de 1 até o número informado.

#Exemplo:

#Digite um número: 5

#Resultado: 15

#Pois:

#1 + 2 + 3 + 4 + 5 = 15

print("---------------SOMA ACUMULADA---------------")

a = int(input("Digite um número inteiro positivo: "))

soma = 0

for i in range(a +1):
  soma += i

print(f"O resultado da soma acumulada é: {soma}")
```

```python
#QUESTÃO 04 – Fatorial (4 pontos)
#----------------------------------------------------------

#Solicite ao usuário um número inteiro positivo e calcule
#seu fatorial utilizando um laço de repetição.

#Exemplo:

#Digite um número: 5

#Fatorial = 120

#Pois:

#5! = 5 × 4 × 3 × 2 × 1

#Observação:
#O programa deve validar se o número informado é positivo.

print("--------------------FATORIAL--------------------")

n = int(input("Digite um número inteiro positivo para o fatorial: "))

f = 1

for i in range(n, 0, -1):
  f *= i

print(f"Fatorial = {f}")
```
